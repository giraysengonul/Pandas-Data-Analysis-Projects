### top 10 highest-rated iphones on flipkart in india?

```python
highest_rated = dataset.sort_values(by=["Star Rating"],ascending=False)
highest_rated = highest_rated.head(10)
```

### how many ratings do the highest-rated iPhones on Flipkart have?

```python
iphones = highest_rated["Product Name"].value_counts()
label = iphones.index
counts = highest_rated["Number Of Ratings"]
figure = px.bar(highest_rated,x=label,y=counts,title="Number of ratings of Highest Rated iPhone")
figure.show()
```

### which iPhone has the highest number of reviews on Flipkart?

```python
iphones = highest_rated["Product Name"].value_counts()
label = iphones.index
counts = highest_rated["Number Of Reviews"]
figure = px.bar(highest_rated,x=label,y=counts,title="Number of reviews | of Highest Rated iPhones")
figure.show()
```

### What is the relationship between the sale price of iPhones and the number of ratings on Flipkart?

```python
dataset.head(1)
```

```python
figure = px.scatter(data_frame=dataset, x="Number Of Ratings", y="Sale Price", size="Discount Percentage",trendline="ols",title="relationship between sales price and number of rating")
figure.show()
```

### What is the relationship between the sale price of iPhones and the number of ratings Flipkart?

```python
figure = px.scatter(data_frame=dataset, 
                    x="Number Of Ratings", 
                    y="Discount Percentage", 
                    size="Sale Price",
                    trendline="ols",
                    title="Relationship between Discount and number of rating")
figure.show()
```

### most expensive and least expensive

```python
most_expensive = dataset.loc[dataset["Sale Price"].idxmax()]
least_expensive = dataset.loc[dataset["Sale Price"].idxmin()]

#Display the results
print("Most Expensive Product:", most_expensive, end="\n.......\n")
print("Least Expensive Product:", least_expensive)
```
